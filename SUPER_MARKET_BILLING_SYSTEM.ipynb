```python
import datetime

# Product list with prices
products = {
    'apple': 200,
    'banana': 100,
    'milk': 150,
    'bread': 250,
    'eggs': 180,
    'rice': 1000
}

cart = {}

# Get the current date and time
current_time = datetime.datetime.now()

print("Welcome to Simple Supermarket Billing System!")

while True:
    print("\n1. Show Products")
    print("2. Add to Cart")
    print("3. Show Bill")
    print("4. Exit")

    choice = input("Choose an option (1-4): ")

    if choice == '1':
        print("\n--- Available Products ---")
        for item, price in products.items():
            print(f"{item.capitalize()} - ${price:.2f}")

    elif choice == '2':
        item = input("Enter product name: ").lower()
        if item in products:
            try:
                quantity = int(input(f"Enter quantity of {item}: "))
                if item in cart:
                    cart[item] += quantity
                else:
                    cart[item] = quantity
                print(f"{quantity} x {item} added to cart.")
            except ValueError:
                print("Invalid quantity. Please enter a number.")
        else:
            print("Product not found.")

    elif choice == '3':
        # Print the bill with "Sekhar Mart" header
        print("\n===== Sekhar Mart =====")
        print(f"Date and Time: {current_time.strftime('%Y-%m-%d %H:%M:%S')}")
        print("------------------------")
        
        total = 0
        for item, qty in cart.items():
            price = products[item] * qty
            print(f"{item.capitalize()} x {qty} = ${price:.2f}")
            total += price
        
        print("------------------------")
        print(f"Total Amount: ${total:.2f}")
        print("========================")

    elif choice == '4':
        print("Thank you for shopping with us!")
        break

    else:
        print("Invalid choice. Please select a valid option.")
```
